```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from datetime import datetime
import os

os.getcwd()
#os.chdir("CSV-Data")
```

```python
viz_nclex = pd.read_csv("merged_vizient_NCLEX.csv")
emp_gpa = pd.read_csv("avg_gpa_by_school.csv")

viz_nclex = viz_nclex.rename(columns={"school_name": "School"})
viz_nclex["School"] = viz_nclex["School"].str.title()

merged_df = pd.merge(viz_nclex, emp_gpa, on="School", how="outer")

merged_df = merged_df[merged_df["year"].notna()]
merged_df = merged_df[merged_df["Average GPA"].notna()]

merged_df.to_csv("merged_viz_nclex_gpa.csv", index=False)
```

```python
hard_skills_df = pd.read_csv("hard_z_scores.csv")
soft_skills_df = pd.read_csv("soft_skills_school_scores.csv")

hard_skills_df = hard_skills_df.rename(columns={"Final Score": "Final Hard Skills Score"})
soft_skills_df = soft_skills_df.rename(columns={"Final score": "Final Soft Skills Score"})

merged_df2 = pd.merge(hard_skills_df, soft_skills_df, on="Schools_Cleaned", how="outer")



merged_df2['Composite Z-Score']             = (
    merged_df2['Final Hard Skills Score'] * 0.50 + 
    merged_df2['Final Soft Skills Score'] * 0.50
)

#merged_df2 = merged_df2[merged_df2['Composite Z-Score'].notna()]

merged_df2.to_csv("merged_skills.csv", index=False)

print(merged_df2['Composite Z-Score'])
```
