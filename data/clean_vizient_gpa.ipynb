```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from datetime import datetime
import os

os.getcwd()
os.chdir("CSV-Data")
```

```python
df = pd.read_csv("DeIdentified_Vizient_Data_New.csv")
df
```

```python
df_clean = df.copy()
df_clean
```

```python

df_clean.columns = (
    df.columns
    .str.strip()
    .str.lower()
)
df_clean
print(df_clean.columns)
```

```python
df_clean["age"] = pd.to_numeric(df_clean["age"], errors="coerce")

# Remove age >= 65 and missing ages

df_clean = df_clean[(df_clean["age"] < 65) & (df_clean["age"].notna())]

# Keep only affiliated schools

valid_schools = {
    "Abraham Baldwin Agricultural Program (Abac)",
    "Albany State University",
    "American Sentinel College",
    "Atlanta Technical College",
    "Auburn University School Of Nursing Main Campus",
    "Augusta University",
    "Baylor University",
    "Brenau University",
    "Capella University",
    "Chamberlain College Of Nursing",
    "Columbus State University",
    "Dalton State College",
    "Duke University",
    "Emory University",
    "Fort Valley State University",
    "Georgia College And State University",
    "Georgia Gwinnett College",
    "Georgia Highlands College",
    "Georgia State University",
    "Gordon State College",
    "Grand Canyon",
    "Herzing University",
    "Indiana University",
    "Jacksonville University",
    "Kennesaw State University",
    "Mercer University",
    "Middle Georgia State",
    "Reinhardt University",
    "South College Atlanta",
    "Thomas University",
    "University Of Alabama Birmingham",
    "University Of Cincinnati",
    "University Of North Georgia",
    "University Of Pittsburgh",
    "University Of South Carolina",
    "University Of Tennessee",
    "University Of West Georgia",
    "Vanderbilt University",
    "Walden University",
    "Western Governor'S University",
    "Xavier University"
}

# normalize for matching
df_clean["education_clean"] = df_clean["education"].str.title()

df_clean = df_clean[df_clean["education_clean"].isin(valid_schools)]

#Remove rows where previous healthcare experience is NOT NaN

df_clean = df_clean[df_clean["previous health care work experience"].isna()]

# Drop column

if "termination updated by" in df_clean.columns:
    df_clean = df_clean.drop(columns=["termination updated by"])
df_clean

if "education" in df_clean.columns:
    df_clean = df_clean.drop(columns=["education"])
df_clean

if "education (cleaned)" in df_clean.columns:
    df_clean = df_clean.drop(columns=["education (cleaned)"])
df_clean
```

```python
df_clean.to_csv("Vizient_CLEANED_FINAL.csv", index=False)

print("Final shape:", df_clean.shape)
print(df_clean.head())
```

```python
gpa_map = {
    "3.5 And Above": 3.75,
    "3.0 - 3.49": 3.25,
    "2.5 - 2.99": 2.75,
    "2.0 - 2.49": 2.25,
    "Below 2.0": 1.75
}

df_clean["gpa"] = (
    df["gpa"]
    .astype(str)
    .str.strip()
    .str.title()
)


df_clean["gpa_numeric"] = df_clean["gpa"].map(gpa_map)

df_clean = df_clean[df_clean["gpa_numeric"].notna()]

avg_gpa_by_school = (
    df_clean
    .groupby("education_clean)")["gpa_numeric"]
    .mean()
    .sort_values(ascending=False)
)

print(avg_gpa_by_school)

avg_gpa_df = avg_gpa_by_school.reset_index()
avg_gpa_df.columns = ["School", "Average GPA"]

print(avg_gpa_df)

counts = df_clean.groupby("education_clean").size()

result = avg_gpa_by_school.to_frame("Average GPA")
result["Count"] = counts

print(result.sort_values("Average GPA", ascending=False))

avg_gpa_df.to_csv("avg_gpa_by_school.csv", index=False)
```
